### Conversation Starter:
In a bulleted list, how is climate change affecting your home? Include references to citations that support your points, and provide a bibliography at the end of this section:

* More exposure to wildfire risk. Hotter, drier conditions make fires more likely and can lengthen the fire season, so we have to think more about defensible buffer space, clearing vegetation, and complying with newer fire-mitigation rules around our homes.  
  Source: https://csfs.colostate.edu/science/colorados-forests-changing-climate/

* Needing more water to keep cultivated landscapes alive. Higher temperatures make water evaporate faster and dry out soil and plants more quickly, even if precipitation stays constant. Which means more irrigation for our yards and farms, more pressure on limited water supplies, and higher water costs.  
  Source: https://nccasc.colorado.edu/NCA5/colorado-water-precipitation-and-drought

* We are spending more energy keeping our homes cool. Hotter summer days mean we rely more on air conditioning, evaporative coolers, and fans during peak heat, which increases electricity use and household energy costs.  
  Source: https://www.colorado.gov/governor/sites/default/files/2023-12/Colorado%20Climate%20Preparedness%20Roadmap_Low%20Resolution%20%281%29.pdf
