### Conversation Starter:

In a bulleted list, how is climate change affecting your home? Include references to citations that support your points, and provide a bibliography at the end of this section:

* **More exposure to wildfire risk.** Hotter, drier conditions make fires more likely and can lengthen the fire season, so we have to think more about defensible space, clearing vegetation, and complying with newer fire-mitigation rules around our homes (Wixson & Cooley, 2023; Colorado Department of Public Safety, 2026).

* **Needing more water to keep cultivated landscapes alive.** Higher temperatures make water evaporate faster and dry out soil and plants more quickly, even if precipitation stays constant. This means more irrigation for our yards and farms, more pressure on limited water supplies, and potentially higher water costs (Milly & Dunne, 2020).

* **We are spending more energy keeping our homes cool.** Hotter summer days mean we rely more on air conditioning, evaporative coolers, and fans during peak heat, which increases electricity use and household energy costs (Amonkar et al., 2023).
